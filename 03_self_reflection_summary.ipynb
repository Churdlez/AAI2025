```python
# Self-Reflection Prompt for Improving Output Goal: Ask the AI to critique and improve a summary.
# Tool used: Google Colab, Gemini, and ChatGPT.
```

```python
## Original Text Prompt chaining uses multiple prompts in sequence. The output from one prompt becomes the input for the next prompt. This makes customer support more organized, but privacy rules and human escalation are still needed for difficult cases.
```

```python
## Self-Reflection Prompt You are an editor reviewing an AI-generated summary.
# Check the summary for: Accuracy, Clarity Missing important information ,Appropriate length, Grammar,Suitability for a general audience

#First provide a short critique. Then write an improved summary between 30 and 40 words.
```

```python
original_summary = """
Prompt chaining uses multiple prompts in sequence to organize customer support.
"""

critique = """
The summary is accurate, but it is incomplete. It does not explain that each
prompt uses the previous output. It also leaves out privacy protections and
human escalation for difficult cases.
"""

improved_summary = """
Prompt chaining connects multiple prompts so each step uses the previous result.
It can organize customer support more effectively, but privacy protections and
human escalation are needed for difficult or sensitive cases.
"""

print("ORIGINAL SUMMARY")
print(original_summary)

print("\nSELF-CRITIQUE")
print(critique)

print("\nIMPROVED SUMMARY")
print(improved_summary)
```
