```python
import sys

class SupportBot:
    def __init__(self):
        # A simple knowledge base
        self.kb = {
            "hours": "We are open from 9 AM to 5 PM, Monday through Friday.",
            "refund": "You can request a refund within 30 days of purchase through your dashboard.",
            "shipping": "Standard shipping takes 3-5 business days. Express is 1-2 days.",
            "human": "I am connecting you to a live agent... (not really, this is a demo!)",
            "greeting": "Hello! How can I help you with your order today?"
        }

    def get_response(self, user_input):
        user_input = user_input.lower()

        # Simple keyword matching logic
        if "hello" in user_input or "hi" in user_input:
            return self.kb["greeting"]
        elif "time" in user_input or "open" in user_input:
            return self.kb["hours"]
        elif "money back" in user_input or "refund" in user_input:
            return self.kb["refund"]
        elif "ship" in user_input or "track" in user_input:
            return self.kb["shipping"]
        elif "help" in user_input or "agent" in user_input:
            return self.kb["human"]
        else:
            return "I'm sorry, I don't understand that. Could you try rephrasing?"

def main():
    bot = SupportBot()
    print("--- Customer Support AI (Type 'exit' to quit) ---")

    while True:
        user_query = input("You: ")
        if user_query.lower() in ["exit", "quit", "bye"]:
            print("Bot: Goodbye! Have a great day.")
            break

        response = bot.get_response(user_query)
        print(f"Bot: {response}")

if __name__ == "__main__":
    main()
```

```python
import time
from IPython.display import clear_output

class ColabSupportBot:
    def __init__(self):
        # The bot's knowledge base
        self.responses = {
            "shipping": "Standard shipping takes 3-5 business days. You'll get a tracking link via email!",
            "refund": "We offer a 30-day money-back guarantee. Just send us your order number.",
            "hours": "Our team is available Monday-Friday, 9 AM to 6 PM EST.",
            "pricing": "Our basic plan starts at $10/month, and the Pro plan is $25/month.",
            "greeting": "Hello! I'm your Support Assistant. How can I help you today?"
        }

    def process_input(self, text):
        text = text.lower()
        if any(word in text for word in ["hi", "hello", "hey"]):
            return self.responses["greeting"]
        elif "ship" in text or "delivery" in text:
            return self.responses["shipping"]
        elif "refund" in text or "return" in text or "money back" in text:
            return self.responses["refund"]
        elif "time" in text or "open" in text or "hours" in text:
            return self.responses["hours"]
        elif "price" in text or "cost" in text or "plan" in text:
            return self.responses["pricing"]
        elif "exit" in text or "bye" in text:
            return "Goodbye! Thanks for reaching out."
        else:
            return "I'm not sure about that. Would you like to speak to a human agent?"

# Initialize the bot
bot = ColabSupportBot()
chat_history = []

print("--- Support Bot Started (Type 'exit' to stop) ---")

while True:
    user_in = input("You: ")

    # Get bot response
    bot_out = bot.process_input(user_in)

    # Store history for a cleaner look
    chat_history.append(f"You: {user_in}")
    chat_history.append(f"Bot: {bot_out}")

    # Refresh the display (Colab specific trick)
    clear_output(wait=True)
    print("--- 💬 Customer Support Chat ---")
    for line in chat_history[-10:]: # Show last 10 messages
        print(line)

    if user_in.lower() in ["exit", "bye", "quit"]:
        break
```

```python
!pip install textblob
from textblob import TextBlob
from IPython.display import clear_output

class EmpatheticBot:
    def __init__(self):
        self.responses = {
            "shipping": "Your order is on its way! Most arrive in 3-5 days.",
            "refund": "I can certainly help with a return. What is your order number?",
            "greeting": "Hello! How can I help you today?"
        }
        self.alert_mode = False

    def get_sentiment(self, text):
        # Returns a value between -1 and 1
        analysis = TextBlob(text)
        return analysis.sentiment.polarity

    def process_input(self, text):
        score = self.get_sentiment(text)
        text = text.lower()

        # 🚨 LOGIC: Check for anger first
        if score < -0.4:
            self.alert_mode = True
            return "⚠️ ALERT: I can tell you're frustrated. I'm prioritizing your case for a human supervisor immediately."

        # Standard keyword matching
        if any(word in text for word in ["hi", "hello"]):
            return self.responses["greeting"]
        elif "ship" in text:
            return self.responses["shipping"]
        elif "refund" in text:
            return self.responses["refund"]
        else:
            return "I'm here to help! Could you give me more details?"

# Initialization
bot = EmpatheticBot()
history = []

print("--- Sentiment-Aware Bot (Try being mean to see the alert) ---")

while True:
    user_in = input("You: ")
    if user_in.lower() in ["exit", "quit"]: break

    bot_out = bot.process_input(user_in)

    # Track history
    history.append(f"You: {user_in}")
    history.append(f"Bot: {bot_out}")

    clear_output(wait=True)
    print("--- 💬 Customer Support Chat ---")
    for line in history[-10:]:
        print(line)

    if bot.alert_mode:
        print("\n[SYSTEM NOTIFICATION]: A manager has been notified of this conversation.")
        break # Stop the bot to prevent further escalation
```
