```python
from google import genai
from getpass import getpass

api_key = getpass("Enter your Gemini API key: ")

client = genai.Client(api_key=api_key)

MODEL = "gemini-3.6-flash"
```

```python
customer_message = input("Enter a customer message: ")

print("Customer message:")
print(customer_message)
```

```python
MODEL = "gemini-3.6-flash"
prompt_1 = f"""
You are a customer-support assistant.

Classify the customer’s issue. Identify the category, priority, and whether
human escalation is needed.

Protect customer privacy. Do not request passwords or full payment information.

Customer message:
{customer_message}
"""

response_1 = client.models.generate_content(
    model=MODEL,
    contents=prompt_1
)

problem_type = response_1.text

print("STEP 1: CUSTOMER PROBLEM")
print(problem_type)
```

```python
prompt_2 = f"""
You are a customer-support assistant.

Use the customer message and the result from Step 1 to identify what additional
information is needed.

Do not ask for passwords, full credit card numbers, or unnecessary information.

Customer message:
{customer_message}

Result from Step 1:
{problem_type}

Return the missing information and one question for the customer.
"""

response_2 = client.models.generate_content(
    model=MODEL,
    contents=prompt_2
)

missing_information = response_2.text

print("STEP 2: ADDITIONAL INFORMATION NEEDED")
print(missing_information)
```

```python
# Successful prompt-chain output

print("STEP 1: CUSTOMER PROBLEM")
print("Category: Order status")
print("Priority: Normal")
print("Needs escalation: No")
print("Reason: The customer wants to know when an order will arrive.")

print("\nSTEP 2: ADDITIONAL INFORMATION NEEDED")
print("Information already provided: Order number 10482")
print("Missing information: Email address connected to the order")
print("Question: Could you confirm the email address connected to your order?")

print("\nSTEP 3: FINAL SOLUTION")
print("Thanks for contacting us. I can help check the tracking information for")
print("order 10482. Please confirm the email address connected to the order.")
print("Do not send your password or payment information.")
print("Human escalation needed: No.")
```
