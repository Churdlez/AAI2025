```python
from google import genai
from getpass import getpass

api_key = getpass("Enter your Gemini API key: ")

client = genai.Client(api_key=api_key)

MODEL = "gemini-3.6-flash"
```

```python
customer_message = input("Enter a customer message: ")

print("Customer message:")
print(customer_message)
```

```python
MODEL = "gemini-3.6-flash"
prompt_1 = f"""
You are a customer-support assistant.

Classify the customer’s issue. Identify the category, priority, and whether
human escalation is needed.

Protect customer privacy. Do not request passwords or full payment information.

Customer message:
{customer_message}
"""

response_1 = client.models.generate_content(
    model=MODEL,
    contents=prompt_1
)

problem_type = response_1.text

print("STEP 1: CUSTOMER PROBLEM")
print(problem_type)
```

```python
prompt_2 = f"""
You are a customer-support assistant.

Use the customer message and the result from Step 1 to identify what additional
information is needed.

Do not ask for passwords, full credit card numbers, or unnecessary information.

Customer message:
{customer_message}

Result from Step 1:
{problem_type}

Return the missing information and one question for the customer.
"""

response_2 = client.models.generate_content(
    model=MODEL,
    contents=prompt_2
)

missing_information = response_2.text

print("STEP 2: ADDITIONAL INFORMATION NEEDED")
print(missing_information)
```

```python
prompt_3 = f"""
You are a polite customer-support agent.

Use the customer message, the Step 1 result, and the Step 2 result to create
a helpful response.

If the issue involves billing, account security, legal concerns, threats,
or a complex problem, recommend human escalation.

Do not request sensitive information.

Customer message:
{customer_message}

Step 1 result:
{problem_type}

Step 2 result:
{missing_information}

Return:
- Recommended action
- Customer response
- Whether human escalation is needed
"""

response_3 = client.models.generate_content(
    model=MODEL,
    contents=prompt_3
)

final_solution = response_3.text

print("STEP 3: FINAL SOLUTION")
print(final_solution)
```

```python
import time

for attempt in range(3):
    try:
        response_3 = client.models.generate_content(
            model=MODEL,
            contents=prompt_3
        )

        final_solution = response_3.text

        print("STEP 3: FINAL SOLUTION")
        print(final_solution)
        break

    except Exception as error:
        print(f"Attempt {attempt + 1} failed.")

        if attempt < 2:
            print("Trying again in 10 seconds...")
            time.sleep(10)
        else:
            print("Gemini is still busy. Please run this cell again later.")
```
