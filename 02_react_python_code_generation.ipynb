```python
## ReACT Prompt

# The AI must first create a short plan. It must then generate Python code, explain how the code should be tested, observe possible errors, and provide a corrected final version.

#The code must calculate the average order value. It must validate that the list is not empty and that all order totals are nonnegative numbers.
```

```python
# ReACT demonstration without an API call

print("PLAN")
print("1. Store the order totals.")
print("2. Validate the list and values.")
print("3. Calculate the average.")
print("4. Print the result with two decimal places.")

print("\nGENERATED CODE")

orders = [35.50, 48.00, 16.50]

if not orders:
    raise ValueError("The order list cannot be empty.")

if any(not isinstance(order, (int, float)) or order < 0 for order in orders):
    raise ValueError("All order totals must be valid nonnegative numbers.")

average_order_value = sum(orders) / len(orders)

print("Code generated successfully.")

print("\nOBSERVATION")
print("The code ran successfully and the input values were valid.")

print("\nFINAL OUTPUT")
print(f"Average order value: ${average_order_value:.2f}")
```
